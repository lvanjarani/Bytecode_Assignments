## 21.	Build Frequency Dictionary
products = [“Laptop”, “Mobile”, “Laptop”, “Tablet”, “Mobile”, “Laptop”]
Create a frequency dictionary using a while loop.


```python
products = ["Laptop", "Mobile", "Laptop", "Tablet", "Mobile", "Laptop"]
freq = {}
i = 0
while i < len(products):
    if products[i] in freq:
        freq[products[i]] =freq[products[i]]+ 1
    else:
        freq[products[i]] = 1
    i = i+1
print(freq)
```

## 24.	Monthly Growth Percentage

## 25.	Mini KPI Dashboard
sales = [1200, 1500, 1800, 2000, 1700]
Using a single while loop, calculate:
•	Total Sales
•	Average Sales
•	Maximum Sales
•	Minimum Sales
•	Number of Records

```python
sales = [1200, 1500, 1800, 2000, 1700]
total = 0
maximum = sales[0]
minimum = sales[0]
count = 0
i = 0
while i < len(sales):
    total =total+ sales[i]
    if sales[i] > maximum:
        maximum = sales[i]
    if sales[i] < minimum:
        minimum = sales[i]
    count =count+ 1
    i =i+ 1
average = total / count
print("Total:", total)
print("Average:", average)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Records:", count)
```

## 30.	Sales Trend Analysis
