Q7. Sort the following list. numbers = [50, 20, 80, 10, 40]
Reverse sort the list.



```python
L2 =[50, 20, 80, 10, 40]
print("Before Sort:",L2)
L2.sort()# ASC
print("After Sort Asc:",L2)
L2.sort(reverse = True)# Desc
print("After Sort Desc:",L2)
```

```python
L2 =[50, 20, 80, 10, 40]
print ("Before Reverse:",L2)
L2.reverse()
print("After Reverse :",L2)
```
