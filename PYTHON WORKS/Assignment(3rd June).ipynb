### Calculate total sales

```python
sales = [1000, 2000, 1500, 3000]

total = 0

for sale in sales:
    total = total + sale

print("Total Sales:", total)
```

### Find highest sales value

```python
sales = [1000, 2000, 1500, 3000]

highest = sales[0]

for sale in sales:
    if sale > highest:
        highest = sale

print("Highest Sales Value:", highest)
```

### Count customers with purchases above 1000

```python
purchases = [500, 1200, 800, 2500, 1500]

count = 0

for amount in purchases:
    if amount > 1000:
        count = count + 1

print("Customers with purchases above 1000:", count)
```
