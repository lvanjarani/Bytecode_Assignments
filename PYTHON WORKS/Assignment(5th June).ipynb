# Question 4
student = {
"name":"Priya",
"age":"25",
"marks":["80","90","85"]
}
- Tasks:
- Convert age into integer
- Convert marks into integers
- Find total and average marks
- Add a new key called "grade"


```python
student = { "name":"Priya", "age":"25", "marks":["80","90","85"] }
student["age"] = int(student["age"])
student["marks"] = [int(mark) for mark in student["marks"]]
total = sum(student["marks"])
average = total / len(student["marks"])
if average >= 90:
    student["grade"] = "A"
elif average >= 75:
    student["grade"] = "B"
elif average >= 50:
    student["grade"] = "C"
else:
    student["grade"] = "Fail"
```

```python
print(student)
print("Total Marks:", total)
print("Average Marks:", average)
```

# Challenge 2
students = {
"Ram":[80,90,85],
"Priya":[95,88,92],
"Hari":[70,75,80]
}
Find:
Total marks of each student
Average marks of each student
Student with highest average


```python
students = {
    "Ram": [80, 90, 85],
    "Priya": [95, 88, 92],
    "Hari": [70, 75, 80]
}

student_averages = {}

for student_name, marks_list in students.items():

    total_marks = 0
    for mark in marks_list:
        total_marks += mark

    average_marks = total_marks / len(marks_list)

    student_averages[student_name] = average_marks

    print(f"\nStudent: {student_name}")
    print(f"Total Marks: {total_marks}")
    print(f"Average Marks: {average_marks:.2f}")

highest_avg_student = max(student_averages, key=student_averages.get)

print(f"\nStudent with Highest Average: {highest_avg_student}")
print(f"Average: {student_averages[highest_avg_student]:.2f}")
```
