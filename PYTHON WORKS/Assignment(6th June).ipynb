# 1. Write a function that returns the largest number from a list without using max().

```python
def largest_number(lst):
    largest = lst[0]
    for i in range(len(lst)):
        if lst[i] > largest:
            largest = lst[i]
    return largest
```

```python
print(largest_number([10, 25, 7, 99, 45]))
```

```python
def Second_Highest_Number(lst):
   largest = lst[0]
   second = lst[0]
   for num in lst:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num
            return second
```

# 5. Find the Separate Even and Odd Numbers : Return two lists.

```python
def separate_even_odd(lst):
    even = []
    odd = []
    for num in lst:
        if num % 2 == 0:
            even.append(num)
        else:
            odd.append(num)
    return even, odd
```

# 6. Find Duplicate Elements : ls = [1,2,3,2,4,5,1]

```python
def find_duplicates(lst):
    duplicate = []
    for num in lst:
        if lst.count(num) > 1 and num not in duplicate:
            duplicate.append(num)
    return duplicate
```

# 10. Reverse Each Word (not string): input: I love python

```python
def reverse_words(sentence):
    words = sentence.split()
    output = ""
    for i in range(len(words)):
        output = output + words[i][::-1] + " "
    return output.strip()
```

# 20. Create Student Grade Dictionary
Input:
students = [
("John",78),
("Priya",92),
("David",45)
]
- Output:
{
'John':'B',
'Priya':'A',
'David':'F'
}
Rules:
A >= 90
B >= 70
C >= 50
F < 50


```python
def student_grades(students):
    grades = {}
    for name, marks in students:
        if marks >= 90:
            grades[name] = "A"
        elif marks >= 70:
            grades[name] = "B"
        elif marks >= 50:
            grades[name] = "C"
        else:
            grades[name] = "F"
    return grades
students = [
    ("John",78),
    ("Priya",92),
    ("David",45)
]
student_grades(students)
```

# 27. Find Pairs Whose Sum = Target
- Input:
nums = [2,7,11,15,3,6]
target = 9
- Output: [(2,7),(3,6)]


```python
def pair_sum(numbers, target):
    pairs = []
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                pairs.append((numbers[i], numbers[j]))
    return pairs
pair_sum([2,7,11,15,3,6], 9)
```

# 28. Compress String
- Input: aaabbccdddd
- Output: a3b2c2d4


```python
def compress(text):
    result = ""
    count = 1
    for i in range(1, len(text)):
        if text[i] == text[i-1]:
            count = count + 1
        else:
            result = result + text[i-1] + str(count)
            count = 1
    result = result + text[-1] + str(count)
    return result
compress("aaabbccdddd")
```

# 29. Expand String
- Input: a3b2c2
- Output: aaabbcc


```python
def expand(text):
    result = ""
    for i in range(0, len(text), 2):
        character = text[i]
        count = int(text[i+1])
        result = result + character * count
    return result
expand("a3b2c2")
```

# 30. Find Consecutive Duplicate Count
- Input: [1,1,1,2,2,3,3,3,3]
- Output:
{
1:3,
2:2,
3:4}

```python
def consecutive_count(numbers):
    result = {}
    current = numbers[0]
    count = 1
    for i in range(1, len(numbers)):
        if numbers[i] == current:
            count = count + 1
        else:
            result[current] = count
            current = numbers[i]
            count = 1
    result[current] = count
    return result
numbers = [1,1,1,2,2,3,3,3,3]
consecutive_count(numbers)
```

# 31. Nested Dictionary Analysis

# 32. Customer Lifetime Value

- Output:
{
'John':4500,
'Priya':8000,
'David':1000
}
- Find:
•	Total spend per customer
•	Highest-value customer
•	Average customer spend



```python
def customer_lifetime_value(customers):
    totals = {}
    highest_customer = ""
    highest_spend = 0
    total_spend = 0
    for customer in customers:
        customer_total = 0
        for amount in customers[customer]:
            customer_total = customer_total + amount
        totals[customer] = customer_total
        total_spend = total_spend + customer_total
        if customer_total > highest_spend:
            highest_spend = customer_total
            highest_customer = customer
    average_spend = total_spend / len(customers)
    print("Customer Totals =", totals)
    print("Highest Customer =", highest_customer)
    print("Average Spend =", average_spend)
customers = {
    "John":[1000,2000,1500],
    "Priya":[5000,3000],
    "David":[1000]
}
customer_lifetime_value(customers)
```

# 33. Build a Mini Dashboard Function
Input:
sales = [1000,2000,5000,3000,7000]
Return:
Total Sales
Average Sales
Highest Sales
Lowest Sales
Sales Count
- output:
in a dictionary:
{
'Total':18000,
'Average':3600,
'Highest':7000,
'Lowest':1000,
'Count':5
}


```python
def dashboard(sales):
    total = 0
    highest = sales[0]
    lowest = sales[0]
    count = len(sales)
    for sale in sales:
        total = total + sale
        if sale > highest:
            highest = sale
        if sale < lowest:
            lowest = sale
    average = total / count
    result = {
        "Total": total,
        "Average": average,
        "Highest": highest,
        "Lowest": lowest,
        "Count": count
    }
    return result
sales = [1000,2000,5000,3000,7000]
dashboard(sales)
```

# 34. Validate Password
- Rules:
Minimum 8 chars
At least 1 uppercase
At least 1 lowercase
At least 1 digit
Return: Valid Or Invalid


```python
def validate_password(password):
    if len(password) < 8:
        return "Invalid"
    has_upper = False
    has_lower = False
    has_digit = False
    for ch in password:
        if ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True
        elif ch.isdigit():
            has_digit = True
    if has_upper and has_lower and has_digit:
        return "Valid"
    return "Invalid"
validate_password("Python123")
```
