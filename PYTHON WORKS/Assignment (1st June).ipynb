5. Customer Analysis Using Sets

february_customers = {103,104,105,106,107}
Find:
•	Repeat *customers*
•	New customers in February
•	Customers who purchased only in January


```python
january_customers = {101,102,103,104,105}
february_customers = {103,104,105,106,107}


print("Repeat Customers :",(january_customers&february_customers))
print("new_customers :",(february_customers - january_customers))
print("january_only:",(january_customers - february_customers))
```

12. Common Products Sold
store1 = {"Laptop","Mouse","Keyboard","Monitor"}
store2 = {"Mouse","Monitor","Printer","Scanner"}
Find:
•	Common products
•	Exclusive products for each store
•	Total unique products

```python
Store1 = {"Laptop","Mouse","Keyboard","Monitor"}
Store2 = {"Mouse","Monitor","Printer","Scanner"}

print("Common_products:",Store1 & Store2)
print("Exclusive_products_store_1:",Store1 - Store2)
print("Exclusive_products_store_2:",Store2 - Store1)
print("Total_unique_products:",Store1 | Store2)
```

13. Mini Case Study
An e-commerce company ran campaigns in 3 months:

jan = {101,102,103,104,105}
feb = {103,104,105,106,107}
mar = {105,106,107,108,109}

Answer:

•	Total unique customers targeted.
•	Customers targeted in all months.
•	Customers targeted only once.
•	Customers targeted in January and February but not March.
•	Customers targeted in February and March but not January.


```python
jan = {101,102,103,104,105}
feb = {103,104,105,106,107}
mar = {105,106,107,108,109}

print("Total_unique:",jan|feb|mar)
print("customers_targeted_all_month:",jan&feb&mar)
print("customers_targeted_Only_once:",(jan | feb | mar) - ((jan & feb) | (jan & mar) | (feb & mar)))
print("customers_targeted_in_january_february:", (jan & feb) - mar)
print("Customers_targeted_February_March:", (feb & mar) - jan)
```
